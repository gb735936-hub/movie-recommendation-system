```python
import ast
```

```python
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name']) 
    return L 

 # these all are in string in which it dont run. to run this you need to put it into list thats why ast is uesd to put in list.   
```

```python
movies['genres'] = movies['genres'].apply(convert)
movies.head()
# genres is change.
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
movies.head()
```

```python
#we need only top 3 values from cast
def convert3(text):
    L = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter < 3:
            L.append(i['name'])
            counter+=1
        else:
            break
    return L 
```

```python
movies['cast'] = movies['cast'].apply(convert3)
movies.head()
```

```python
#crew=we need the director from all the crew.
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
            break
    return L 
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
movies.head()
```

```python
#overview is in string so convert it in list.
movies['overview'] = movies['overview'].apply(lambda x:x.split())
movies.head()
```

```python
#'Sam Worthington'
#'SamWorthington' we need remove a space between one name because model think its two different name
# so we do that in last four column.
```

```python
movies['genres'] = movies['genres'].apply(
    lambda x: [i.replace(" ", "") for i in x]
)

movies['keywords'] = movies['keywords'].apply(
    lambda x: [i.replace(" ", "") for i in x]
)

movies['cast'] = movies['cast'].apply(
    lambda x: [i.replace(" ", "") for i in x]
)

movies['crew'] = movies['crew'].apply(
    lambda x: [i.replace(" ", "") for i in x]
)
movies.head()
```

```python
# now we concatinate the 4 column in one =tags
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
movies.head()
```

```python
#now we create a new dataframe
new_df=movies[['movie_id','title','tags']]
```

```python
new_df
```

```python
#now convert list into string of tags
new_df['tags'] = new_df['tags'].apply(lambda x: " ".join(x))
new_df.head()
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps= PorterStemmer()
```

```python
def stem(text):
    y=[]

    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)    
```

```python
new_df['tags']= new_df['tags'].apply(stem)
```

```python
new_df['tags'][0]
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x: x.lower())
new_df.head()
#convert in lowercase
```

```python
#'''text= we convert now text to vector and we see the similarites between these tags and the closer tags is recommodation'''


```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vector = cv.fit_transform(new_df['tags']).toarray()
```

```python
# we calculate the distance between the cosine angle between 
#the one tag to another tag. if angle is closer the sililar distance is less and movies is similar 
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity.shape
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list= sorted(list(enumerate(distances)),reverse=True,key = lambda x: x[1])[1:6]
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
        
```

```python
recommend('Avatar')
```
